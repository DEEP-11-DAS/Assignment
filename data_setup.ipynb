```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_olivetti_faces
```

```python
seed = 26004
rng = np.random.default_rng(seed)
```

```python
my_data = fetch_olivetti_faces()
```

```python
x = my_data.data
y = my_data.target
```

```python
train = []
val = []
test = []
id = 0

while id < 40:
    start_idx = id * 10
    person_idx = []
    for i in range(10):
        person_idx.append(i + start_idx)

    shuffled_idx = rng.permutation(person_idx)

    train.extend(shuffled_idx[0:6])
    val.extend(shuffled_idx[6:8])
    test.extend(shuffled_idx[8:10])

    id += 1
```

```python
# making them actual arrays
X_train = x[train]
X_val = x[val]
X_test = x[test]

y_train = y[train]
y_val = y[val]
y_test = y[test]
```

```python
print("train shape:", X_train.shape)
```

```python
temp = X_test.copy()
temp1 = temp.reshape(80, 64, 64)
```

```python
mask = np.ones((64, 64))
mask[24:40, 24:40] = 0.0
```

```python

X_test_masked = (temp1 * mask).reshape(len(temp), -1)
```

```python
np.savez_compressed(
    "dat_split.npz",
    X_train=X_train,
    X_val=X_val,
    X_test_clean=X_test,
    X_test_masked=X_test_masked,
    y_train=y_train,
    y_val=y_val,
    y_test=y_test
)
print("saved to dat_split.npz!")
```
